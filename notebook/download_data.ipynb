2. Processing wind data and mapping to satellite images
3. Maintaining temporal alignment
4. Saving prepared data for deep learning

```python
import os
import glob
import zipfile
from datetime import datetime, timedelta
import numpy as np
import pandas as pd
from satpy import Scene
from pyresample import create_area_def
from pyresample.geometry import AreaDefinition
import matplotlib.pyplot as plt
from PIL import Image
import pdbufr
from pathlib import Path
import json

# Create output directories
os.makedirs('processed_data', exist_ok=True)
os.makedirs('processed_data/satellite_images', exist_ok=True)
os.makedirs('processed_data/wind_data', exist_ok=True)
os.makedirs('processed_data/combined', exist_ok=True)

print("✅ Libraries imported and directories created")
```

```python
# Define Morocco Region of Interest (ROI)
# Morocco Bounding Box coordinates
MOROCCO_ROI = {
    'lat_min': 21.0,
    'lat_max': 36.0,
    'lon_min': -17.0,
    'lon_max': -1.0,
    'resolution': 3000  # 3km resolution
}

# Create pyresample area definition for Morocco
morocco_area = create_area_def(
    'morocco',
    {'proj': 'latlong', 'datum': 'WGS84'},
    area_extent=[MOROCCO_ROI['lon_min'], MOROCCO_ROI['lat_min'], 
                 MOROCCO_ROI['lon_max'], MOROCCO_ROI['lat_max']],
    resolution=MOROCCO_ROI['resolution'],
    units='meters',
    description='Morocco ROI'
)

print(f"Morocco ROI defined:")
print(f"  Latitude: {MOROCCO_ROI['lat_min']}° to {MOROCCO_ROI['lat_max']}°")
print(f"  Longitude: {MOROCCO_ROI['lon_min']}° to {MOROCCO_ROI['lon_max']}°")
print(f"  Resolution: {MOROCCO_ROI['resolution']}m")
print(f"  Area shape: {morocco_area.shape}")
```

```python
# Function to extract timestamp from filename
def extract_timestamp_from_filename(filename):
    """
    Extract timestamp from satellite or wind filenames
    MSG4-SEVI-MSG15-0100-NA-20200303002743.637000000Z-NA.zip
    W_XX-EUMETSAT-Darmstadt,VIS_IR_WV+ATMOSPHERE+HRW,MSG2+SEVIRI_C_EUMG_20200331T090000Z_20200331T091000Z_1_OR_RSS_E0950_0100
    """
    try:
        if 'MSG4-SEVI' in filename or 'MSG' in filename:
            # Extract from MSG satellite filename
            # Format: ...20200303002743...
            import re
            match = re.search(r'(\d{8})T?(\d{6})', filename)
            if match:
                date_str = match.group(1)
                time_str = match.group(2)
                timestamp = datetime.strptime(date_str + time_str, '%Y%m%d%H%M%S')
                return timestamp
        elif 'W_XX' in filename:
            # Extract from wind filename
            # Format: ...20200331T090000Z...
            import re
            match = re.search(r'(\d{8})T(\d{6})Z', filename)
            if match:
                date_str = match.group(1)
                time_str = match.group(2)
                timestamp = datetime.strptime(date_str + time_str, '%Y%m%d%H%M%S')
                return timestamp
    except Exception as e:
        print(f"Error extracting timestamp from {filename}: {e}")
    return None

# Test the function
test_files = [
    'MSG4-SEVI-MSG15-0100-NA-20200303002743.637000000Z-NA.zip',
    'W_XX-EUMETSAT-Darmstadt,VIS_IR_WV+ATMOSPHERE+HRW,MSG2+SEVIRI_C_EUMG_20200331T090000Z_20200331T091000Z_1_OR_RSS_E0950_0100'
]
for f in test_files:
    ts = extract_timestamp_from_filename(f)
    print(f"File: {f[:50]}...")
    print(f"Timestamp: {ts}\n")
```

```python
# Function to process satellite image and extract Morocco ROI
def process_satellite_image(zip_path, output_dir='processed_data/satellite_images'):
    """
    Process a satellite image zip file:
    1. Extract the .nat file
    2. Load with satpy
    3. Crop to Morocco ROI
    4. Save as numpy array and image
    """
    try:
        print(f"Processing: {os.path.basename(zip_path)}")
        
        # Extract timestamp
        timestamp = extract_timestamp_from_filename(zip_path)
        if timestamp is None:
            print(f"  ⚠️ Could not extract timestamp, skipping")
            return None
        
        # Extract zip file to temporary directory
        temp_dir = 'temp_extract'
        os.makedirs(temp_dir, exist_ok=True)
        
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(temp_dir)
        
        # Find .nat file
        nat_files = glob.glob(os.path.join(temp_dir, '*.nat'))
        if not nat_files:
            print(f"  ⚠️ No .nat file found")
            return None
        
        # Load scene
        scn = Scene(filenames=nat_files, reader='seviri_l1b_native')
        
        # Load multiple channels (you can adjust based on your needs)
        channels = ['IR_108', 'IR_120', 'VIS006', 'VIS008']
        available_channels = []
        
        for ch in channels:
            try:
                scn.load([ch])
                available_channels.append(ch)
            except:
                pass
        
        if not available_channels:
            print(f"  ⚠️ No channels could be loaded")
            return None
        
        # Resample to Morocco ROI
        scn_resampled = scn.resample(morocco_area)
        
        # Extract data as numpy arrays
        processed_data = {}
        for ch in available_channels:
            data = scn_resampled[ch].values
            processed_data[ch] = data
        
        # Save processed data
        timestamp_str = timestamp.strftime('%Y%m%d_%H%M%S')
        output_file = os.path.join(output_dir, f'sat_{timestamp_str}.npz')
        np.savez_compressed(output_file, **processed_data, timestamp=timestamp_str)
        
        # Save visualization
        if 'IR_108' in processed_data:
            img_data = processed_data['IR_108']
            # Normalize for visualization
            img_norm = (img_data - np.nanmin(img_data)) / (np.nanmax(img_data) - np.nanmin(img_data))
            img_norm = (img_norm * 255).astype(np.uint8)
            img = Image.fromarray(img_norm)
            img_file = os.path.join(output_dir, f'sat_{timestamp_str}.png')
            img.save(img_file)
        
        # Clean up temp files
        for f in glob.glob(os.path.join(temp_dir, '*')):
            os.remove(f)
        
        print(f"  ✅ Saved: {output_file}")
        
        return {
            'timestamp': timestamp,
            'file': output_file,
            'channels': available_channels,
            'shape': processed_data[available_channels[0]].shape
        }
        
    except Exception as e:
        print(f"  ❌ Error: {str(e)}")
        return None

print("✅ Satellite processing function defined")
```

```python
# Function to process wind data and extract Morocco ROI
def process_wind_data(bufr_file, output_dir='processed_data/wind_data'):
    """
    Process a wind BUFR file:
    1. Read wind data
    2. Filter for Morocco ROI
    3. Create gridded wind field
    4. Save as numpy array
    """
    try:
        print(f"Processing: {os.path.basename(bufr_file)}")
        
        # Extract timestamp
        timestamp = extract_timestamp_from_filename(bufr_file)
        if timestamp is None:
            print(f"  ⚠️ Could not extract timestamp, skipping")
            return None
        
        # Read BUFR file
        df = pdbufr.read_bufr(
            bufr_file,
            columns=[
                "latitude",
                "longitude",
                "windSpeed",
                "windDirection",
                "pressure",
                "data_datetime"
            ]
        )
        
        # Filter for Morocco region
        morocco_wind = df[
            (df["latitude"] >= MOROCCO_ROI['lat_min']) & 
            (df["latitude"] <= MOROCCO_ROI['lat_max']) &
            (df["longitude"] >= MOROCCO_ROI['lon_min']) & 
            (df["longitude"] <= MOROCCO_ROI['lon_max'])
        ]
        
        print(f"  Found {len(morocco_wind)} wind vectors in Morocco ROI")
        
        if len(morocco_wind) == 0:
            print(f"  ⚠️ No wind data in ROI")
            return None
        
        # Convert wind direction and speed to U and V components
        # U = -speed * sin(direction), V = -speed * cos(direction)
        morocco_wind = morocco_wind.copy()
        morocco_wind['wind_u'] = -morocco_wind['windSpeed'] * np.sin(np.radians(morocco_wind['windDirection']))
        morocco_wind['wind_v'] = -morocco_wind['windSpeed'] * np.cos(np.radians(morocco_wind['windDirection']))
        
        # Create gridded wind field
        # Define grid
        lat_bins = np.linspace(MOROCCO_ROI['lat_min'], MOROCCO_ROI['lat_max'], 50)
        lon_bins = np.linspace(MOROCCO_ROI['lon_min'], MOROCCO_ROI['lon_max'], 50)
        
        # Grid wind data using 2D histogram weighted average
        from scipy.ndimage import generic_filter
        
        # Create grids for U and V components
        wind_u_grid = np.zeros((len(lat_bins)-1, len(lon_bins)-1))
        wind_v_grid = np.zeros((len(lat_bins)-1, len(lon_bins)-1))
        wind_speed_grid = np.zeros((len(lat_bins)-1, len(lon_bins)-1))
        count_grid = np.zeros((len(lat_bins)-1, len(lon_bins)-1))
        
        for _, row in morocco_wind.iterrows():
            lat_idx = np.digitize(row['latitude'], lat_bins) - 1
            lon_idx = np.digitize(row['longitude'], lon_bins) - 1
            
            if 0 <= lat_idx < len(lat_bins)-1 and 0 <= lon_idx < len(lon_bins)-1:
                wind_u_grid[lat_idx, lon_idx] += row['wind_u']
                wind_v_grid[lat_idx, lon_idx] += row['wind_v']
                wind_speed_grid[lat_idx, lon_idx] += row['windSpeed']
                count_grid[lat_idx, lon_idx] += 1
        
        # Average the grids
        mask = count_grid > 0
        wind_u_grid[mask] /= count_grid[mask]
        wind_v_grid[mask] /= count_grid[mask]
        wind_speed_grid[mask] /= count_grid[mask]
        
        # Save processed data
        timestamp_str = timestamp.strftime('%Y%m%d_%H%M%S')
        output_file = os.path.join(output_dir, f'wind_{timestamp_str}.npz')
        
        np.savez_compressed(
            output_file,
            wind_u=wind_u_grid,
            wind_v=wind_v_grid,
            wind_speed=wind_speed_grid,
            count=count_grid,
            lat_bins=lat_bins,
            lon_bins=lon_bins,
            timestamp=timestamp_str,
            raw_data=morocco_wind.to_dict()
        )
        
        print(f"  ✅ Saved: {output_file}")
        
        return {
            'timestamp': timestamp,
            'file': output_file,
            'n_vectors': len(morocco_wind),
            'shape': wind_u_grid.shape
        }
        
    except Exception as e:
        print(f"  ❌ Error: {str(e)}")
        import traceback
        traceback.print_exc()
        return None

print("✅ Wind processing function defined")
```

```python
# Temporal alignment and matching
print("=" * 60)
print("TEMPORAL ALIGNMENT")
print("=" * 60)

# Create dataframes for easier matching
sat_df = pd.DataFrame(satellite_processed)
wind_df = pd.DataFrame(wind_processed)

if len(sat_df) > 0 and len(wind_df) > 0:
    sat_df = sat_df.sort_values('timestamp')
    wind_df = wind_df.sort_values('timestamp')
    
    print(f"\nSatellite images: {len(sat_df)}")
    print(f"  Time range: {sat_df['timestamp'].min()} to {sat_df['timestamp'].max()}")
    
    print(f"\nWind data: {len(wind_df)}")
    print(f"  Time range: {wind_df['timestamp'].min()} to {wind_df['timestamp'].max()}")
    
    # Match satellite and wind data by timestamp (within 15 minutes tolerance)
    matched_pairs = []
    time_tolerance = timedelta(minutes=15)
    
    for _, sat_row in sat_df.iterrows():
        sat_time = sat_row['timestamp']
        
        # Find closest wind measurement
        time_diffs = abs(wind_df['timestamp'] - sat_time)
        closest_idx = time_diffs.idxmin()
        
        if time_diffs[closest_idx] <= time_tolerance:
            wind_row = wind_df.loc[closest_idx]
            matched_pairs.append({
                'timestamp': sat_time,
                'satellite_file': sat_row['file'],
                'wind_file': wind_row['file'],
                'time_diff_seconds': time_diffs[closest_idx].total_seconds()
            })
    
    print(f"\n✅ Matched {len(matched_pairs)} satellite-wind pairs")
    
    # Save matching metadata
    matched_df = pd.DataFrame(matched_pairs)
    matched_df.to_csv('processed_data/matched_pairs.csv', index=False)
    print("✅ Saved matching pairs to processed_data/matched_pairs.csv")
else:
    print("⚠️ No data to match")
    matched_pairs = []
```

```python
# Create combined dataset for deep learning
print("=" * 60)
print("CREATING COMBINED DATASET")
print("=" * 60)

from scipy.ndimage import zoom

for idx, pair in enumerate(matched_pairs):
    try:
        # Load satellite data
        sat_data = np.load(pair['satellite_file'])
        
        # Load wind data
        wind_data = np.load(pair['wind_file'])
        
        # Get the primary satellite channel (IR_108)
        sat_channels = [k for k in sat_data.keys() if k not in ['timestamp']]
        if not sat_channels:
            continue
            
        primary_channel = sat_channels[0] if 'IR_108' not in sat_channels else 'IR_108'
        sat_image = sat_data[primary_channel]
        
        # Resize wind grids to match satellite image dimensions
        wind_u = wind_data['wind_u']
        wind_v = wind_data['wind_v']
        wind_speed = wind_data['wind_speed']
        
        # Calculate zoom factors
        zoom_factor_y = sat_image.shape[0] / wind_u.shape[0]
        zoom_factor_x = sat_image.shape[1] / wind_u.shape[1]
        
        # Resize wind data
        wind_u_resized = zoom(wind_u, (zoom_factor_y, zoom_factor_x), order=1)
        wind_v_resized = zoom(wind_v, (zoom_factor_y, zoom_factor_x), order=1)
        wind_speed_resized = zoom(wind_speed, (zoom_factor_y, zoom_factor_x), order=1)
        
        # Stack all channels together
        # Shape: (channels, height, width)
        combined = np.stack([
            sat_image,
            wind_u_resized,
            wind_v_resized,
            wind_speed_resized
        ], axis=0)
        
        # Save combined data
        timestamp_str = pair['timestamp'].strftime('%Y%m%d_%H%M%S')
        output_file = f'processed_data/combined/combined_{timestamp_str}.npz'
        
        np.savez_compressed(
            output_file,
            data=combined,
            timestamp=timestamp_str,
            channels=['satellite', 'wind_u', 'wind_v', 'wind_speed']
        )
        
        print(f"✅ [{idx+1}/{len(matched_pairs)}] Saved {output_file}")
        print(f"   Shape: {combined.shape}")
        
    except Exception as e:
        print(f"❌ Error processing pair {idx}: {e}")

print(f"\n✅ Combined dataset created in processed_data/combined/")
```

```python
# Create sequences for time series forecasting
print("=" * 60)
print("CREATING TEMPORAL SEQUENCES")
print("=" * 60)

# Load all combined files and sort by timestamp
combined_files = sorted(glob.glob('processed_data/combined/combined_*.npz'))

print(f"Found {len(combined_files)} combined files")

# Create sequences (e.g., use 6 timesteps to predict next timestep)
SEQUENCE_LENGTH = 6  # Number of past timesteps to use
PREDICTION_HORIZON = 1  # How many steps ahead to predict

sequences = []

for i in range(len(combined_files) - SEQUENCE_LENGTH - PREDICTION_HORIZON + 1):
    # Input sequence
    input_files = combined_files[i:i+SEQUENCE_LENGTH]
    
    # Target (what we want to predict)
    target_file = combined_files[i+SEQUENCE_LENGTH]
    
    # Load input sequence
    input_data = []
    timestamps = []
    
    for f in input_files:
        data = np.load(f)
        input_data.append(data['data'])
        timestamps.append(data['timestamp'])
    
    # Load target
    target_data = np.load(target_file)
    
    # Stack into sequence
    # Shape: (sequence_length, channels, height, width)
    input_sequence = np.stack(input_data, axis=0)
    target = target_data['data']
    
    sequences.append({
        'input_shape': input_sequence.shape,
        'target_shape': target.shape,
        'start_time': timestamps[0],
        'end_time': timestamps[-1],
        'target_time': target_data['timestamp']
    })

print(f"\n✅ Created {len(sequences)} sequences")
print(f"   Sequence length: {SEQUENCE_LENGTH}")
print(f"   Input shape: {sequences[0]['input_shape'] if sequences else 'N/A'}")
print(f"   Target shape: {sequences[0]['target_shape'] if sequences else 'N/A'}")

# Save sequence metadata
if sequences:
    seq_df = pd.DataFrame(sequences)
    seq_df.to_csv('processed_data/sequences_metadata.csv', index=False)
    print("✅ Saved sequence metadata to processed_data/sequences_metadata.csv")
```

## Data Loading Helper for Deep Learning

Use this code to load the prepared data in your deep learning model:

```python
class MoroccoWeatherDataset:
    """
    Dataset class for loading prepared Morocco weather data
    Compatible with PyTorch and TensorFlow
    """
    def __init__(self, data_dir='processed_data/combined', sequence_length=6):
        self.data_dir = data_dir
        self.sequence_length = sequence_length
        
        # Load all files
        self.files = sorted(glob.glob(os.path.join(data_dir, 'combined_*.npz')))
        print(f"Found {len(self.files)} data files")
        
        # Create sequences
        self.sequences = []
        for i in range(len(self.files) - sequence_length):
            self.sequences.append({
                'input_files': self.files[i:i+sequence_length],
                'target_file': self.files[i+sequence_length]
            })
        
        print(f"Created {len(self.sequences)} sequences")
    
    def __len__(self):
        return len(self.sequences)
    
    def __getitem__(self, idx):
        """
        Returns:
            X: Input sequence of shape (sequence_length, channels, height, width)
            y: Target of shape (channels, height, width)
        """
        seq = self.sequences[idx]
        
        # Load input sequence
        X = []
        for f in seq['input_files']:
            data = np.load(f)['data']
            X.append(data)
        X = np.stack(X, axis=0)  # (seq_len, channels, H, W)
        
        # Load target
        y = np.load(seq['target_file'])['data']  # (channels, H, W)
        
        return X, y
    
    def get_sample(self, idx=0):
        """Get a sample for visualization"""
        return self[idx]

# Example usage:
if len(combined_files) > 0:
    dataset = MoroccoWeatherDataset('processed_data/combined', sequence_length=6)
    
    # Get a sample
    X, y = dataset.get_sample(0)
    print(f"\nDataset sample shapes:")
    print(f"  Input (X): {X.shape}  # (sequence_length, channels, height, width)")
    print(f"  Target (y): {y.shape}  # (channels, height, width)")
    
    print(f"\n✅ Dataset class ready for PyTorch/TensorFlow")
else:
    print("⚠️ No data available to create dataset")
```
